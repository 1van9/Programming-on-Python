## Бонусное задание 2. Deadline tracker

В данном задании нужно написать программу в отдельном файле, которая работает с дедлайнами через аргументы командной строки. Программа должна уметь добавлять дедлайны, изменять их и определять надвигающиеся дедлайны.

**Требования от программы:**
- Пользователь должен иметь возможность добавить новый дедлайн (у дедлайна есть два параметра, время (в идеале уметь переводить время из разных форматов, но можно и остановиться на формате dd.mm.yyyy) и название).

- Пользователь должен иметь возможность поменять дату или название уже имеющегося дедлайна.

- Также пользователь может удалить имеющийся дедлайн.

- Суть приложения - программа должна отвечать, по данной дате, какие дедлайны нужно сделать к указанной дате, а какие дедлайны только надвигаются (нужно сделать на следующий день и через два дня после указанной даты).

**Пример запуска:**

Программа должна запускаться из консоли, например так (названия для программы и команд могу быть другими):

```bash
python deadline_tracker.py add 17.12.2026 "Сдать бонусное задание по информатике"
```

Как назвать программу, команды к ней, а также в каком формате выводить информацию о дедлайнах - решать вам. Главное, чтобы было читаемо и понятно.

**Указания к решению:**

1) Вашу программу будут запускать многократно, любые переменные, которые вы сохраните в программе удалятся, как только она завершится, однако данные о дедлайнах сохранять надо. Делать это предлагается в отдельном файле. Можете разобраться с форматом `json`, но в данном задании подойдут обычные тхтшки. 

2) Вам потребуется работать с нестандартным для вас запуском программы через консоль. И данные, которые будут передаваться в консоль не получится считать через `input`. Чтобы получить данные, необходимо воспользоваться библиотекой `sys`.

    `sys.argv` возвращает список строк, которые написаны в командной строке после python имя_файла.py.

    Например программа `sum.py`:

    ```python
    import sys

    print(int(sys.argv[1]) + int(sys.argv[2]))
    ```

    при запуске:

    ```bash
    python sum.py 2 2
    ```
    Будет печатать 4, а в `sys.argv` будет лежать список `['sum.py', '2', '2']`. 
    
    Python разделяет аргументы по пробелу. Если хочется в один аргумет положить несколько слов, то можно эти слова передать в кавычках, например

    ```bash
    python code.py "hello world!"
    ```

    Тогда у `sys.argv` будет всего два элемента в списке.

    Все элементы `sys.argv` будут строчками, нулевой элемент - всегда название программы.
