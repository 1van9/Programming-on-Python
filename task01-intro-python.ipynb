### Тесты к задаче 3.

```python
test_cases_quadratic_roots = [
    ([1, 2, 1], [-1]),
    ([1, 2, 2], []),
    ([1, -2, -2], [-0.732050807568, 2.732050807568]),
    ([3, 0, 0], [0]),
    ([1, -2, 0], [0, 2]),
    ([7, -4, 2], []),
    ([7, 4, -2], [-0.891805812445, 0.320377241017]),
    ([13, 72, -45], [-6.105422837261, 0.566961298800])
]

def check_roots(value, result):
    if not isinstance(value, list):
        return False
    if len(value) != len(result):
        return False
    v = sorted(value)
    r = sorted(result)
    for i in range(len(v)):
        if abs(v[i] - r[i]) > 0.000001:
            return False
    return True

for (args, result) in test_cases_quadratic_roots:
    value = quadratic_roots(*args)
    if not check_roots(value, result):
        print("Wrong answer, arg: " + 
        ", ".join(map(str, args)) + ", expected: " + str(result) +
        ", but returned: " + str(value))
        break
else:
    print("Problem 3. Tests passed")
```
