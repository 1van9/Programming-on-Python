Что же касается округления: есть 4 способа округлять

- `int(x)` - отбрасывает дробную часть
- `round(x)` - округляет до ближайшего целого
- `math.floor(x)` - округляет вниз
- `math.ceil(x)` - округляет ввех

Для последних двух нужна библиотека math.

Разница int и math.floor в том, что для отрицательных чисел int округляет наверх, а math.floor - вниз.

```python
import math
a = 12.34
b = -5.7
c = -3.21
d = 1.5

print("value   int floor  ceil round")
print(f"{a:5}", f"{int(a):5}", f"{math.floor(a):5}", f"{math.ceil(a):5}", f"{round(a):5}")
print(f"{b:5}", f"{int(b):5}", f"{math.floor(b):5}", f"{math.ceil(b):5}", f"{round(b):5}")
print(f"{c:5}", f"{int(c):5}", f"{math.floor(c):5}", f"{math.ceil(c):5}", f"{round(c):5}")
print(f"{d:5}", f"{int(d):5}", f"{math.floor(d):5}", f"{math.ceil(d):5}", f"{round(d):5}")
```

## Объекты

Мы уже сказали, что функция `range()` возвращает объект определенного типа (типа range). Однако хорошо бы понимать, что значат **объект**, **тип**, а также **переменная** и **значение**. Давайте разбираться на примере:

```python
a = 5
```

В данном случае, мы создали **переменную** `a` - **объект** **типа** `int` со **значением** 5.

В нашей программе мы хотим работать с какими-то данными, данные эти хранятся в нашем компьютере, непосредственно в микрочипах и накопителях по которым бежит ток. *Объект*, это непосредственно область в памяти, отвечающая за данные, которые мы храним. Сами данные, которые хранятся в этой памяти, это *значение*. В данном случае значение 5, которое лежит где-то на компьютере, записанное в виде ноликов и единичек. *Переменная* же, по сути не хранит значение напрямую, она запоминает только адрес памяти (номера ячеек, в которых данный объект лежит). Адрес в памяти еще называют **ссылкой**. Ну а *тип* помогает компьютеру распарсить, как правильно относиться к объекту: как к числу или символу, а может быть как к целому массиву, ведь в памяти копьютера хранятся только нолики и единички.

Если вы никогда не думали про то, где хранятся данные вашей программы, то вероятно это ломает мозг, но постепенно к этому нужно привыкнуть.

## Функции

Напишем пример простой функции

```python
def power(a, k):
    result = 1
    while k > 0:
        result *= a
        k -= 1
    return result
```

```python
base = 2
exp = 5
power(base, exp)
```

Слово `def` обозначает, что перед нами функция, `a` и `k` ее аргументы, а команда `return` возвращает данные, которые будут вычесленны после запуска этой функции с определенными аргументами (в данном случае `base` и `exp`).

К каждому элементу можно обратиться по индексу. Вспомним как работает индексация на примере

```python
#       a = [1, 2, 3, 4, 5]
# indexes:   0  1  2  3  4
# also:     -5 -4 -3 -2 -1
```

Так например есть изысканный способ пробежаться по массиву:

```python
a = [1, 2, 3, 4]

for i in range(0, len(a)):
    print(a[i], end = " ")
print()

for i in range(-len(a), 0):
    print(a[i], end = " ")
print()
```

При обращении по другим индексам возникнет всем известный `list index out of range`, поэтому внимательно следите по каким индексам вы обращаетесь.

Срезы. Работают по аналогии с range




```python
a = [0, 1, 2, 3, 4, 5]
print("a[2:4] =", a[2:4])
print("a[2::2] =", a[2::2])
print("a[::3] =", a[::3])
print("a[:-4:-1] =", a[:-4:-1])
print("a[-1:-6:-1] =", a[-1:-6:-1])
print("a[::-1] =", a[::-1])
```

Рассмотрим основные функции и методы для работы со списками:

- `len(a)` - длина списка `a`
- `a.append(x)` - добавить в конец списка `a` элемент `x`
- `a.insert(i, x)` - добавить на позицию `i` элемент `x`
- `a.extend(b)` - добавить в конец списка `a` элементы списка `b`
- `a.pop()` - удалить последний элемент списка и возвращает его (ошибка, если список пустой). Возвращает удаленный элемент
- `a.pop(i)` - удалить элемент с индексом `i` (в списке должен быть элемент с таким индексом). Возвращает удаленный элемент
- `a.remove(x)` - удалить элемент со значением `x` (в списке должен быть элемент со значением `x`)
- `a.clear()` - удалить все элементы списка
- `a.count(x)` - найти количество вхождений элемента `x` в списке
- `a.index(x)` - найти в списке индекс элемента равного `x` (в списке должен быть такой элемент)
- `max(a), min(a), sum(a)` - несложно определить из названия (max и min не работают с пустыми списками)
- `a.reverse()` - разворернуть список `a`, изменяя его
- `reversed(a)` - возвраща развернутый список не меняя `a`
- `a.sort()` - сортирует список
- `sorted(a)` - возвращает остортированный список

Все операции за исключением `len()`, `append(x)` и `pop()` (которые выполняются быстро) работают за проход по массиву, а сортировка вообще за $O(n \log n)$, где $n$ - размер массива.

Для лучшего понимания ознакомьтесь с примерами

```python
a = [1, 2, 3]
print("len(a) =", len(a))
a.append(57)
print("afret append(57) a =", a)
a.insert(0, 100)
print("after insert(0, 100) a =", a)
a.extend([58, 58, 59])
print("after extend([58, 58, 59]) a =", a)
a.pop()
print("after pop() a =", a)
a.pop(2)
print("after pop(2) a =", a)
a.remove(58)
print("after remove(58) a =", a)
a.extend([57, 58])
print("a =", a)
print("a.count(-1) =", a.count(-1))
print("a.count(57) =", a.count(57))
print("a.index(58) =", a.index(58))
print("min(a) =", min(a))
print("max(a) =", max(a))
print("sum(a) =", sum(a))
print("reversed(a) =", list(reversed(a)))
print("sorted(a) =", sorted(a))
a.clear()
print("after clear() a =", a)
```
