## Задание 6. Получи TL

Рассмотрим классическую задачу о рюкзаке: Дано `n` предметов, у каждого предмета есть вес `weight` и цена `value`, а также есть рюкзак, выдерживающий вес `capacity`. Требуется набрать несколько предметов так, чтобы их суммарный вес не превосходил `capacity`, а суммарная ценность была максимальна.

У данной задачи есть переборное решение, которое асимптотически работает за $O(2^{n})$ - перебор всех возможных вариантов, однако на практике оно работает сильно быстрее из-за отсечений заведомо неоптимальных вариантов. Приведем это решение:  

```python
def greedy_knapsack(items, capacity):
    best_value = 0
    taken = []
    remaining = capacity
    for i in range(len(items)):
        weight, value = items[i]
        if weight <= remaining:
            remaining -= weight
            best_value += value
            taken.append(i)
    return best_value, taken


def fractional_bound(items, start, remaining):
    bound = 0.0
    for i in range(start, len(items)):
        weight, value = items[i]
        if weight <= remaining:
            remaining -= weight
            bound += value
        else:
            bound += value * (remaining / weight)
            break
    return bound


def knapsack_branch_and_bound(items, capacity):
    n = len(items)
    if n > 150 or n == 0:
        raise ValueError("invalid data size")
    if any(map(lambda x: x[0] <= 0 or x[1] <= 0, items)):
        raise ValueError("weights and values of items must be positive")
    items.sort(key=lambda x: x[1] / x[0], reverse=True)
    best_value, best_taken = greedy_knapsack(items, capacity)
    cur_taken = []

    def dfs(index, cur_value, cur_weight):
        nonlocal best_value, best_taken

        if cur_value > best_value:
            best_value = cur_value
            best_taken = cur_taken.copy()

        if index == n:
            return

        upper_bound = cur_value + fractional_bound(items, index, capacity - cur_weight)
        if upper_bound <= best_value:
            return

        weight, value = items[index]
        if cur_weight + weight <= capacity:
            cur_taken.append(index)
            dfs(index + 1, cur_value + value, cur_weight + weight)
            cur_taken.pop()

        upper_bound = cur_value + fractional_bound(items, index + 1, capacity - cur_weight)
        if upper_bound > best_value:
            dfs(index + 1, cur_value, cur_weight)

    dfs(0, 0, 0)
    return best_value, sorted(best_taken)
```
