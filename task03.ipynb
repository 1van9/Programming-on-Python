## Функция тестирования программ

Не забудьте запустить данные ячейки

```python
import threading
import time
import math
import random
from itertools import permutations
```

```python
def short(message):
    if len(message) < 50:
        return message
    return message[:20] + '...'

def runner(res, solution, args):
    try:
        value = solution(*args)
        res[0] = (True, value)
    except BaseException as e:
        res[0] = (False, f"{type(e).__name__}: {e}")

def tester(problem, solution, checker, tests, maxTime=2):
    for (args, result) in tests:
        res = [None]
        solve = threading.Thread(target = runner, args=(res, solution, args, ))
        solve.start()
        solve.join(timeout=maxTime)

        if solve.is_alive():
            print(f"{maxTime} second time limit exceeded.")
            break

        if not res[0][0]:
            print(res[0][1])
            break
        if not checker(res[0][1], result):
            print("Wrong answer, arg: " +
            short(", ".join(map(str, args))) + ", expected: " + short(str(result)) + ", but returned: " + short(str(res[0][1])) + ".")
            break
    else:
        print(f"Problem {problem}. Tests passed")
```

Хеш-функция должна стараться возвращать разные значения для разных списков, иначе `set` будет работать долго.

```python
class MyList:
    def __init__(self, x):
        self.arr = x
    def __hash__(self):
        # your code here
        pass
    def __eq__(self, other):
        return isinstance(other, MyList) and self.arr == other.arr
    def __repr__(self):
        return f"{self.arr}"
```

### Тест

```python
def test_hash_list(arrays_to_add, arrays_to_pop, arrays_in, arrays_out):
    s = set()
    for i in range(len(arrays_to_add)):
        s.add(MyList(arrays_to_add[i]))
        assert len(s) == i + 1
    for i in range(len(arrays_to_add)):
        s.add(MyList(arrays_to_add[i]))
        assert len(s) == len(arrays_to_add)
    for i in range(len(arrays_to_add)):
        assert MyList(arrays_to_add[i]) in s
    for i in range(len(arrays_to_pop)):
        s.remove(MyList(arrays_to_pop[i]))
        assert len(s) == len(arrays_to_add) - i - 1
    for arr in arrays_in:
        assert MyList(arr) in s
    for arr in arrays_out:
        assert MyList(arr) not in s
    return 0

perms = list(permutations(range(1, 9)))
erase_perms = [x for x in perms if x[3] == 1]
arrays_in = [x for x in perms if x[3] != 1]
arrays_out = erase_perms + list(permutations(range(8)))

test_cases_hash_list = [
    (([[1], [2], [3]], [[2], [1]], [[3]], [[2], [1], [-1], [0], [4]]), 0),
    (([[1], [1, 2], [1, 2, 3]], [[1, 2, 3]], [[1], [1, 2]], [[2, 1], [3, 2, 1], [1, 2, 3]]), 0),
    (([[1, 2, 3], [3, 2, 1], [2, 3, 1], [1, 3, 2], [4]], [[1, 3, 2], [4]], [[1, 2, 3], [3, 2, 1]], [[1, 3, 2], [3, 1, 2]]), 0),
    ((perms, erase_perms, arrays_in, arrays_out), 0),
    (([[i for i in range(j)] for j in range(1, 1000)], [[i for i in range(j)] for j in range(1, 1000, 2)],
      [[i for i in range(j)] for j in range(2, 1000, 2)], [[i for i in range(j - 1, -1, -1)] for j in range(2, 1000)]), 0)
]

tester(12, test_hash_list, lambda x, y: 1, test_cases_hash_list)
```
