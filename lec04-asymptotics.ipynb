# Асимптотика

Часто, когда мы пишем код, хочется оценить, насколько он эффективный и как быстро будет выполняться. О том, как принято об этом думать будет в этой лекции.

## Измерение скорости кода

Очевидной метрикой было бы посчитать, сколько времени будет выполняться код: чем быстрее, тем лучше. Однако один и тот же код может быть запущен и на новейшем компьютере, напичканном мощным процессором, топовой видеокартой, сотнями гигабайт оперативной памяти и разными ускорителями, и на бабушкиной микроволновке 90-х годов. А нам бы хотелось универсальный способ оценки программ. Поэтому вместо времени мы будем считать количество элементарных операций выоплняемых программой.

Элементарные операции: присваивание, сравнение, арифметика, обращение к элементу массива.

Данные можно параметризовать. Для этого выбирают параметры, обычно количество данных $n$, диапозоны значений $C$ и др. в зависимости от задачи. И количество действий программых оценивается не на конкретных данных, а на произвольных данных с фиксированными параметрами, как функция от этих параметров.

### Пример

```python
def sum_array(arr):
    res = 0              # 1 операция
    for x in arr:        # n итераций
        res += x         # 1 операция на итерацию
    return res           # 1 операция
```

В данном случае нашим параметром выступил размер массива - $n$.

Считаем действия:
- `res = 0` - $1$ действие
- `res += x` - $n$ действий
- `return res` - $1$ действие

Итого: $f(n) = n + 2$

Асимптотика обычно оценивается по худшему случаю — максимальному числу операций среди всех входов с заданым параметром.

```python
def contains(arr, target):
    for x in arr:
        if x == target:
            return True     # может найтись сразу - O(1)
    return False            # а может и не быть   - O(n)
```

- Лучший случай: target в начале - $1$ операция.
- Худший случай: target отсутствует - $n$ операций.

Асимптотика - $O(n)$, потому что мы всегда берём худший случай.

**Важно:** константы и младшие слагаемые можно отбрасывать, что значительно упрощает подсчет числа операций. $n + 2$ и $100n + 500$ - оба $O(n)$. Поговорим чуть подробнее, что это значит.

## О большое    

Нам не важно, во сколько раз алгоритм A быстрее алгоритма B на маленьких данных. Нам важно, как растёт время при увеличении $n$.

- $n + 2$ и $100n + 500$ растут линейно - оба $O(n)$.
- $n^2$ и $5n^2 + 6n - 3$ растут квадратично - оба $O(n^2)$.

O-большое — это «класс роста»: мы говорим, что $f(n) = O(g(n))$, если функция $f(n)$ не больше чем $C \cdot g(n)$ при больших $n$ для некоторой константы $C$.

Обычно в качестве оценки асимптотики используются простые функции:

- $O(1)$ -	    константа
- $O(\log n)$ - растёт очень медленно
- $O(n)$ - линейный рост
- $O(n \log n)$ или $O(n \log C)$ - чуть хуже линейного
- $O(n^2)$ - 	квадратичный рост
- $O(2^n)$ - 	экспоненциальный рост

В качестве оценки программы используется **наименьшая** возможная асимптотика в худшем случае. То есть, если программа выполняет $3n + 5$ действий, конечно, верно будет сказать, что она работает за $O(n^5)$, ведь $n^5 > 3n + 5$ при больших $n$, однако в данном случае правильно будет написать асимптотику $O(n)$.

## Примеры

#### Пример 1

```python
def quadratic_roots(a, b, c):
    assert a != 0
    b /= 2
    D = b * b - a * c
    if D < 0:
        return []
    if D == 0:
        return [-b / a]
    return [(-b - D**0.5) / a, (-b + D**0.5) / a]
```

Операций константное число - асимптотика $O(1)$.

#### Пример 2

```python
def count_even(arr):
    cnt = 0
    for x in arr:          # n итераций
        if x % 2 == 0:     # 1 сравнение
            cnt += 1       # иногда +1
    return cnt
```

Количество операций зависит от входных данных, но независимо от этого, их будет не более чем $2n + 2$, то есть асимптотика $O(n)$.

#### Пример 3

```python
def sum_of_two(arr, target):
    for i in range(len(arr)):          # n итераций
        j = i + 1
        while j < len(arr):            # не более n итераций
            if arr[i] + arr[j] == target:
                return (i, j)
            j += 1
    return ()
```

**Худший случай:** подходящей пары нет (или пара в самом конце, на позициях $n - 2$, $n - 1$) - внешний цикл сделает $n$ операций, внутренний будет делать $n - j$ операций, то есть суммарно $n(n - 1) / 2$ операций, что равно $O(n^2)$.

#### Пример 4

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True
```

Цикл выполняется не более чем $\sqrt{n} - 1$ раз, в худшем случае, для простого $n$ ровно $\lfloor \sqrt{n} \rfloor - 1$, таким образом асимптотика $O(\sqrt{n})$. 

#### Пример 5

```python
def upper_power(n):
    k = 0
    while 2**k < n:
        k += 1
    return k
```

Цикл `while` будет выполняться $k$ действий, итоговое $k = \lceil \log_2 n \rceil$, то есть асимптотика $O(\log n)$. Однако есть нюанс, сама по себе операция `**` не всегда выполняется за $O(1)$, если $n$ очень большое, состоит из тысячи цифр, то возведение в степень $\log n$ будет работать за $\log (\log n)$, кроме того операции с такими большими числами тоже перестанут работать за единицу, но более подробно мы этого касаться не будем.

Для большей надежности можно избавиться от возведения, добившись почти чистой асимптотики $O(\log n)$:

```python
def upper_power(n):
    k = 0
    while n > 1:
        k += 1
        n = (n + 1) / 2
    return k
```
