## Файлы


Файл - это последовательность байт сохранненая в компьютере. Файлы имеют имя `name` и расширение `ext`, тогда в компьютере это будет файл `name.ext`, расширение позволяет понять компьютеру, как интерпретировать данные : как текст (расширение `txt`), как изображение (расширения `jpg`, `png`), как аудиозапись (расширение `mp3`), как программу на языке python (расширение `py`), и наконец непосредственно как последовательность байт (расширение `bin`). Бывают и другие расширения.

Файлы на копьютере хранятся в папочках, и всегда есть путь до файла `C:\Users\ivang\prog\code.cpp` - последовательность папочек, чтобы добраться до файла (разделитель `\` используется в Windows, `/` в Mac и Linux). Путь от самого начала - с диска - называется **абсолютным**, но обычно мы уже находимся в каком-то файле, например запускаем программу из какого-то файла, тогда можно рассмотреть **относительный** путь `..\..\..\ivang\prog`, `..\` означает, что надо выйти из текущей папки.

Чтобы работать с файлом в `python`, его сначала нужно **открыть**. Для этого используется функция `open()`. Она создает дескриптор, который уже умеет читать данные из файла и записывать их обратно.

После окончания работы с файлом его нужно закрыть с помощью `close()`. Если не написать `close()`, то часть переданных программой данных могу быть потеряны. Более того часто при открытии файла операционная система блокирует его, останавливая доступ других программ к этому файлу. Однако, когда программа завершается, дескриптор автоматически закрывает файл.

````python
f = open("numbers.txt", "w", encoding="utf-8")
f.write("1 2 3 4 5")
f.close()
````

Здесь `"numbers.txt"` — имя файла, `"w"` — режим открытия файла. Также можно передать относительный или абсолютный путь до файла.

````python
f = open(".\\folder\\file.txt") # открыть файл из папки folder
f.close()
````

Параметр `encoding="utf-8"` задает кодировку файла. Это особенно важно, если в файле используется русский текст. (По умолчанию на Linux и Mac кодировка `utf-8`, однако на Windows используется `cp1251`, из-за чего данные будут считываться некорректно).

````python
f1 = open("russian.txt", "r", encoding="utf-8")
print(f1.read())
f2 = open("russian.txt", "r", encoding = "cp1251")
print(f2.read())
````

### Режимы открытия файла

| Режим  | Что делает                                               |
| ------ | -------------------------------------------------------- |
| `'r'`  | открыть файл для чтения                                  |
| `'w'`  | открыть файл для записи, старое содержимое будет удалено |
| `'a'`  | открыть файл для записи в конец                          |
| `'r+'` | открыть файл для чтения и записи                         |

По умлочанию (если не передавать режим) файл будет открываться в режиме чтения.

Если открыть несуществующий файл в режиме `'r'`, возникнет ошибка.

Если открыть несуществующий файл в режиме `'w'` или `'a'`, Python создаст его.

Например, при каждом таком запуске старое содержимое файла будет удаляться:

````python
f = open("data.txt", "w", encoding="utf-8")
f.write("Hello")
f.close()

f = open("data.txt", "w", encoding="utf-8")
f.write("World")
f.close()
````

В итоге в файле будет только:

```text
World
```

Если же использовать режим `'a'`, текст будет добавляться в конец файла:

````python
f = open("data.txt", "a", encoding="utf-8")
f.write("!") # Hello!
f.close()
````

**Упражнение:** Что произойдет с содержимым файла, если открыть его сначала с режимом `'w'`, а затем с режимом `'a'`?

### Конструкция `with open(...)`

Закрывать файл вручную с помощью `close()` не всегда удобно. Например, во время работы с файлом может произойти ошибка, и программа не дойдет до `close()`.

Поэтому обычно используют конструкцию `with`:

````python
with open("data.txt", "w", encoding="utf-8") as f:
    f.write("Hello,\n world!")
````

После выхода из блока `with` файл автоматически закрывается.

Это безопаснее и удобнее, поэтому при обычной работе с файлами лучше использовать именно такой вариант.

### Чтение из файла

Для чтения всего содержимого файла используется метод `read()`, который возвращает строку, записанную в файле (с учетом всех пробелов и переносов строк).

````python
with open("data.txt", "r") as f:
    print(f.read())
````

Можно указать количество символов, которые нужно прочитать: `f.read(5)`

Если в файле было:

```text
Hello,
 world!
```

то программа прочитает первые 5 символов:

```text
Hello
```

Функция `readline()` читает одну строку (аналогично `readline(x)` считывает только `x` символов строки):

````python
with open("data.txt", "r") as f:
    print(f.readline())
````

Следующий вызов `readline()` прочитает следующую строку:

````python
with open("data.txt", "r", encoding="utf-8") as f:
    print(f.readline())
    print(f.readline())
````

Функция `readlines()` читает все строки и возвращает их в виде списка:

````python
with open("data.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

print(lines)
````

Например, если файл содержит:

```text
one
two
three
```

то результат будет примерно таким:

```python
['one\n', 'two\n', 'three']
```

Обратите внимание на `\n`. Это специальный символ конца строки. Чтобы его убрать, можно использовать метод `line.strip()`. `strip` убирает и пробелы, стоящие в начале и конце строки.

### Перебор файла по строкам


Файл можно перебирать в цикле `for`:

````python
with open("data.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())
````

По сути в `line` считываются строки по одной. Такой способ лучше чем `for line in f.readlines()` особенно для больших файлов: не обязательно загружать весь файл в память сразу, а обрабатывать его по одно й строке.

### Запись в файл

Для записи используется функция `write()`:

````python
with open("numbers.txt", "w", encoding="utf-8") as f:
    f.write("1")
    f.write("2")
    f.write("3")
````

Обратите внимание: `write()` не добавляет перевод строки автоматически как это делает `print`.

````python
with open("numbers.txt", "w", encoding="utf-8") as f:
    f.write("1\n")
    f.write("2\n")
    f.write("3\n")
````

Теперь каждое число находится на отдельной строке.

Важно понимать, что `write` записывает текст "поверх" файла (заменяя уже существующие символы), а не в конец его (чтобы дописывать в конец файла нужен режим `'a'`).

Функция `writelines()` записывает несколько строк:

````python
numbers = ["1\n", "2\n", "3\n"]

with open("numbers.txt", "w", encoding="utf-8") as f:
    f.writelines(numbers)
````

Также для записи в файл можно использовать `print()`, возможно это более привычно. Нужно передать ему параметр `file`, в этом случае `print()` работает не с экраном, а с файлом.

````python
with open("numbers.txt", "w", encoding="utf-8") as f:
    print(1, file=f)
    print(2, file=f)
    print(3, file=f)
````

### Курсор

Когда мы пишем `readline()`, или `read(x)` мы считываем какую-то часть файла и следующие команды будут читать файл с того момента остановки. И чтобы вернуться к началу или считать что-то повторно, конечно можно снова открыть файл, считать сколько-то символов из начала, выкинуть их и вот, пожалуйста.

Но это слишком трудоемкий процесс. Вместо этого можно просто сдвинуть позицию чтения файла с помощью `f.seek(k)`. После этого `read()` будет читать файл с `k + 1`-го символа. Чтобы узнать на каком символе сейчас находится курсор, есть метод `f.tell()`, он возвращает позицию курсора.

````python
with open("data.txt", "r", encoding="utf-8") as f:
    print(f.tell())  # 0
    f.read(3)
    print(f.tell())  # 3
    f.seek(1)
    print(f.tell(), f.read(4))  # 1 ello
````

При работе с текстовыми файлами в Python позиции курсора не всегда соответствуют количеству символов. На самом деле она соответствует количеству байт от начала файла: это особенно важно при использовании разных кодировок, когда некоторые символы занимают больше чем 1 байт. Для обычных текстовых файлов в режиме чтения можно пользоваться приведенными выше примерами.

Методы `seek()` и `tell()` также полезны при записи. Например, можно переместить курсор и перезаписать часть файла:

````python
with open("data.txt", "r+", encoding="utf-8") as f:
    print(f.read())
    f.seek(0)
    f.write("XYZ")
    f.seek(0)
    print(f.read())
````

На самом деле у метода `seek` есть второй параметр:

`f.seek(offset, whence)`:
- `offset` — смещение, на которое нужно переместить курсор.
- `whence` — точка отсчёта, относительно которой вычисляется смещение.

Если `whence = 0`, то курсор смещается от начала файла, если `whence = 1`, то от текущего положения, если `whence = 2`, то от конца (да, `offset` может быть отрицательным).

В случае если `whence` отличен от нуля, смещать курсор можно только в бинарном режиме `'rb'`\ `'r+b'` \ `'wb'` \ `'ab'` (как последовательность бит). В таком случае, кодировки никакой нет, чтобы по считанным битам получить строчку, нужно воспользоваться методом `decode`. Аналогично, чтобы записать строчку, нужно перевести ее в последовательнось бит: `s.encode("utf-8")`.

````python
with open("data.txt", "r+b") as f:
    f.seek(-2, 2)
    print(f.read().decode())
    f.seek(-2, 2)
    f.write(":)".encode())
    f.seek(0)
    print(f.read().decode())
````

## Консоль

Обычно мы взаимодействуем с компьютером через графический интерфейс: открываем папки, нажимаем кнопки, запускаем программы мышью. Но многие действия можно выполнять и с помощью текстовых команд.

В Windows для этого можно использовать **командную строку** (cmd) или **PowerShell**. В Linux обычно используют терминал.

Командная строка особенно полезна при работе с файлами, запуске программ, обработке больших объёмов данных и автоматизации задач.

Командная строка (она же терминал) при запуске открывается в некоторой папке и имеет непосредственный доступ к файлам в ней. Чтобы посмотреть содержимое текущей папки, в Windows можно выполнить `dir`, а в Linux — аналогичную команду `ls`.

### Консольные команды

Команды зависят от операционной системы и используемой оболочки, поэтому синтаксис иногда различается.

| Действие | Windows (cmd) | Linux (Bash) |
|---|---|---|
| Показать текущую папку | `cd` | `pwd` |
| Посмотреть содержимое папки | `dir` | `ls` |
| Перейти в папку | `cd имя_папки` | `cd имя_папки` |
| Перейти на уровень выше | `cd ..` | `cd ..` |
| Перейти в корень текущего диска или файловой системы | `cd \` | `cd /` |
| Создать папку | `mkdir test` | `mkdir test` |
| Удалить пустую папку | `rmdir test` | `rmdir test` |
| Удалить файл | `del a.txt` | `rm a.txt` |
| Скопировать файл | `copy a.txt b.txt` | `cp a.txt b.txt` |
| Переместить или переименовать файл | `move a.txt b.txt` | `mv a.txt b.txt` |
| Вывести текст | `echo Hello` | `echo Hello` |
| Очистить экран | `cls` | `clear` |
| Вывести содержимое текстового файла | `type a.txt` | `cat a.txt` |
| Найти текст в файле | `find "hello" a.txt` | `grep "hello" a.txt` |
| Показать различия	файлов | `fc a.txt b.txt` | `diff a.txt b.txt` |
| Побайтовое сравнение файлов | `comp a.txt b.txt` | `cmp a.txt b.txt` |
| Создать файл с текстом |`echo Hello > a.txt` |	`echo "Hello" > a.txt` |
| Дописать текст в файл |`echo Hello >> a.txt` |	`echo "Hello" >> a.txt` |

Чтобы командная строка вносила какие-то существенные изменения в компьютер, например устонавливала программы, нужно выдать ей на это доступ, для этого можно запустить командную строку от имени администратора в Windows. В Linux, нужно перед командой поставить `sudo` (она запросит пароль), например `sudo apt install gcc`.

Из Python можно запускать внешние команды с помощью стандартного модуля subprocess.

Например, посмотрим, какая в гугл коллабе версия python:

````python
import subprocess

result = subprocess.run(
    ["python", "--version"],
    capture_output=True,
    text=True
)

print(result.stdout)
print(result.stderr)
print(result.returncode)
````

Здесь:

- `subprocess.run(...)` запускает внешнюю программу;
- список `["python", "--version"]` задаёт программу и её аргументы (то что будет запущено в командной строке);
- `capture_output=True` сохраняет вывод программы (будет положен в поле .stdout)
- `text=True` позволяет получить вывод в виде обычных строк, а не бинарных (.stdout будет строкой)
- `.stderr` - вывод ошибок программы (ошибки обычно выводятся в отдельный поток, просто пользователь видит их также как и вывод).
- `returncode` - код завершения программы. Обычно 0 означает успешное выполнение.
