# Сортировки

Сортировка располагает список чисел по порядку. Но чтобы ей пользоваться этот порядок нужно задать. 

## Упорядоченные множества

### Отношение порядка
На множестве можно задать **отношение порядка** - правило, которое говорит, что один элемент «не больше» («меньше») другого, обозначается $\leqslant$ (или $<$).

Чтобы это отношение вело себя осмысленно, оно должно подчиняться аксиомам:

- Рефлексивность: $a \leqslant a$ — каждый элемент не больше самого себя.
- Антисимметричность: если $a \leqslant b$ и $b \leqslant a$, то $a = b$.
- Транзитивность: если $a \leqslant b$ и $b \leqslant c$, то $a \leqslant c$.

Если выполняются все три - перед нами корректный порядок и на нем можно строить сортировку.

#### Линейный порядок

Отношение порядка задает **линейный порядок**, если любая пара элементов сравнима, т. е. для любых $a$ и $b$ верно либо $a \leqslant b$, либо $b \leqslant a$.

Конечное множество с линейным порядком всегда можно упорядочить так, чтобы $$a_1 \leqslant a_2 \leqslant  \dots \leqslant a_{n - 1}\leqslant a_n,$$ это и будет прямой порядок сортировки.

**Примеры прямого порядка:**

- Числа с обычным сравнением.
- Строки с лексикографическим сравнением.
- Массивы с поэлементным сравнением.
- Кортежи одинаковой длины с покомпонентным сравнением.
- Даты, время.

#### Частичный порядок

Частичный порядок - это когда не все пары сравнимы. Некоторые элементы «несравнимы»: ни $a \leqslant b$, ни $b \leqslant a$ не выполняется.

**Примеры частичного порядка:**
- Делимость натуральных чисел: 
    - $2 \leqslant 6$ ($2$ делит $6$), 
    - $5 \leqslant 5$ ($5$ делит $5$), 
    - $4 \not\leqslant 6$ и $6 \not\leqslant 4$, (4 и 6 друг друга не делят).
- Множества с отношением вложенности: $A \leq B \Leftrightarrow A \subseteq B$.


*Упражнение:* Для всех предложенных примеров линейного и частичного порядка, докажите, что они являются отношениями порядка.

Для конечного множества с заданным частичным порядком можно расположить числа так, чтобы выполнялось условие: если $a \leqslant b$, то $a$ идет раньше чем $b$. Такая сортировка называется топологической.

## Параметры сортировки

Для основных типов задано отношение порядка (определены операции `<`, `>`, `<=`, `>=`). Как уже было сказано раньше, строчки сортируются лексикографически, массивы и кортежи поэлементно:
- `5 <= 8`
- `'hello' < 'hi'`
- `[1, 2, 3] < [2]`
- `(1, 2, 3) > (1, 2)`

Сортировка по умолчанию сравнивает элементы исходя из этого порядка сравнения, она не пользуется представлениями самих объектов, но сортировке необходимо уметь сравнивать элементы.

Например, сравнивать разные типы нельзя: `[1, 2, 3] < '123'` - TypeError. Поэтому сортируемый массив должен состоять из одинаковых попарно сравниваемых объектов.

#### Сортировка по ключу

На практике иногда нужно сортировать не «сами элементы», а какой-то их признак:

- по длине строки,
- по возрасту человека,
- по модулю числа,
- по последней цифре.

**Ключ** - это функция, которая каждому элементу ставит в соответствие значение для сравнения. Сортировка происходит по ключам, а не по самим элементам. Ключ должен возвращать значения, сравнимые между собой.

Если нужно несколько критериев — ключ возвращает кортеж, а кортежи сравниваются лексикографически (сначала по первому, при равенстве — по второму, и т.д.). Так «сортировка по нескольким полям» сводится к линейному порядку на кортежах.

```python
string_arr = ['ab', 'c', 'a', 'de']
string_arr.sort(key = len) # string_arr = ['c', 'a', 'ab', 'de']
different_types = [1, 'abc', 0.5, [4]]
different_types.sort(key = lambda x : str(type(x))) # diffrent_types = [0.5, 1, [4], 'abc']
print(string_arr, different_types)                  #           'float', 'int', 'list', 'string'
```

#### Компаратор

**Компаратор** - это функция от двух элементов, которая их сравнивает. Обычно компаратор должен возвращать `True` т. и т. т., к. один элемент **строго меньше** чем другой. 

Иногда компаратор устроен по другому и возвращает:
- $-1$, если $a < b$, 
- $1$, если $a > b$,
- $0$, если $a = b$.

В любом случае на основе компаратора сортировка сравнивает элементы.

Чтобы сортировка работала корректно, компаратор должен удовлетворять условиям отношения порядка:
- **Антирефлективность:** $a = b$ т. и т. т. к. не верно, что $a < b$ и $b < a$.
- **Антисимметричность:** если $a < b$, то неверно, что $b < a$.
- **Транзитивность:** если $a < b$ и $b < c$, то $a < c$.

```python
from functools import cmp_to_key

def fraction_compare(a, b):
    if a[0] * b[1] < a[1] * b[0]:
        return -1
    if a[0] * b[1] > a[1] * b[0]:
        return 1
    return 0

fracs = [(2, 5), (1, 2), (1, 3), (7, 9)]
fracs.sort(key = cmp_to_key(fraction_compare)) # 1/3 < 2/5 < 1/2 < 7/9
print(fracs) # [(1, 3), (2, 5), (1, 2), (7, 9)]
```

#### Стабильность

**Стабильная сортировка** - это сортировка, которая сохраняет относительный порядок равных элементов.

То есть если $a$ и $b$ имеют одинаковый ключ (или равны с точки зрения компаратора), и $a$ стоял раньше $b$ до сортировки, то после сортировки $a$ тоже будет раньше $b$.



Если сортировка стабильна, то можно запустить многоуровневую сортировку через несколько проходов.
Можно сортировать не одним сложным ключом, а последовательно: сначала по вторичному признаку,  потом по первичному. Устойчивость гарантирует, что вторичный порядок сохранится внутри групп с одинаковым первичным ключом.

Так, например, сначала можно отсортировать по последней цифре чисел, потом стабильно по предпоследней, затем по разряду сотен и так далее. Таким образом получится поразрядная сортировка. 

*Упражнение:* докажите, что поразрядная сортировка действительно сортирует числа.

Применив идею стабильной сортировки подсчетом, и выполняя разделения по цифрам в $n$-ичной системе счисления можно сортировать числа за $O(n \log_n{C})$, что лучше стандартной сортировки за $O(n \log n)$ однако стандартная сортировка более универсальна.

**Важно:** если пользоваться только сравнениями элементов при сортировке, то отсортировать элементы быстрее чем за $O(n \log n)$ нельзя.
