### Тест задачи 6

Генератор тестов c аномалиями (аномалии генерируются не всегда корректно, так как при их добавлении меняется среднее и стандартное отклонение)

```python
import random

def generate_large_test(n, k, mu, sigma, seed=None):
    if seed is not None:
        random.seed(seed)
    
    clean = [random.gauss(mu, sigma) for _ in range(n)]
    anomalies = [mu + random.choice([1, -1]) * random.uniform(3 * sigma, 100 * sigma) for _ in range(k)]
    data = clean + anomalies
    random.shuffle(data)
    return ([data], anomalies)
```
