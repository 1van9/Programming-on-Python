## Функция тестирования программ

Не забудьте запустить данные ячейки

```python
import copy
from itertools import permutations
import math
import random
import time
import threading

def read_intarr(f):
    return list(map(int, f.readline().strip().split()))

def read_arr(f):
    return f.readline().strip().split()

def read_string(f):
    return f.readline().strip()
```

```python
def short(message):
    if len(message) < 50:
        return message
    return message[:20] + '...'

def runner(res, solution, args):
    try:
        value = solution(*args)
        res[0] = (True, (*args, value))
    except BaseException as e:
        res[0] = (False, f"{type(e).__name__}: {e}")

class test_result:
    def __init__(self, status, message):
        self.status = status
        self.message = message

    def __add__(self, other):
        if self.status:
            return other
        return self

    def __repr__(self):
        return self.message

    def __bool__(self):
        return self.status
    

def tester(problem, solution, checker, tests, maxTime=2):
    for (args, result) in tests:
        res = [None]
        inp = copy.deepcopy(args)
        solve = threading.Thread(target = runner, args=(res, solution, args, ))
        solve.start()
        solve.join(timeout=maxTime)

        if solve.is_alive():
            return test_result(False, f"{maxTime} second time limit exceeded.")

        if not res[0][0]:
            return test_result(False, res[0][1])
        if not checker(*res[0][1], result):
            return test_result(False, "Wrong answer, args: " + short(", ".join(map(str, inp))))
    else:
        return test_result(True, f"Problem {problem}. Tests passed")


def file_tester(problem, problem_name, solution, indata, outdata, N, checker, maxTime=2):
    for i in range(1, N + 1):
        res = [None]
        args = indata(open(f"{problem_name}/{i}.in", "r"))
        solve = threading.Thread(target = runner, args=(res, solution, args, ))
        solve.start()
        solve.join(timeout=maxTime)

        if solve.is_alive():
            return test_result(False, f"{maxTime} second time limit exceeded on test {i}.")

        if not res[0][0]:
            return test_result(False, f"On test {i}:" + res[0][1])
        expected = outdata(open(f"{problem_name}/{i}.out", "r"))
        if not checker(*res[0][1], expected):
            return test_result(False, f"Wrong answer on test {i}.")
    else:
        return test_result(True, f"Problem {problem}. Tests passed")
```
