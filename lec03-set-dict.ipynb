# Set и Dict

## Set

**set** - это буквально множество - неупорядоченная коллекция уникальных элементов:

```python
fruits = {"яблоко", "банан", "апельсин"}
print("fruits:", fruits)

nums = {1, 2, 2, 3, 3, 3}
print("nums:", nums)
```

Создать множество из других объектов можно через функцию `set()` или через генерацию по аналогии со списками:

```python
empty_set = set()
set_from_list = set([1, 3, 1, 3, 2])
set_from_string = set('Hello')
generated_set = {x for x in range(10) if x % 2 == 0}
print(empty_set, set_from_list, set_from_string, generated_set)
```

Операции с множествами:

| Операция | Что делает |
|----------|------------|
|len(s)| кол-во элементов в `s`|
|s.add(x)| добавить в `s` элемент `x`|
|s.update(l)| добавить в `s` коллекцию элементов `l`|
|s.remove(x)| удалить `x` из множества (Если `x` не было, то будет ошибка)|
|s.discard(x)| безопасно удалить `x`|
|s.pop()| удалить случайный элемент, его значение возвращается |

```python
s = {1, 2, 3}
s.add(4)          # {1, 2, 3, 4}
s.add(2)          # {1, 2, 3, 4}
s.update([5, 6])  # {1, 2, 3, 4, 5, 6}

s.remove(3)       # {1, 2, 4, 5, 6}
# повторно s.remove(3) - ошибка 
s.discard(100)    # {1, 2, 4, 5, 6}
popped = s.pop()
print("delete", popped, "from {1, 2, 4, 5, 6},", "result:", s)
s.clear()         # set()
```

Математические операции с множествами:

| Операция | Что делает |
|----------|------------|
|`a \| b` / a.union(b) | объединить два множества $a \cup b$ |
|`a & b` / a.intersection(b) | пересечь два множества $a \cap b$|
|`a - b` / a.difference(b)| разность множеств $a \setminus b$|
|`a ^ b` / a.symmetric_difference(b) | симметрическая разность $a \Delta b$|
|`a = b`| проверка, что множества состоят из одинаковых элементов $a = b$|
|`a < b`| проверка, что одно множество вложено в другое, но не равно $a \subsetneq b$|
| `a <= b` / a.issubset(b)|проверка, что одно множество вложено в другое $a \subset b$|
|`a > b`, `a >= b` / a.issuperset(b)| аналогично |
|a.isdisjoint(b)| проверка, что множества не пересекаются $a \cap b = \emptyset$|

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(a | b) # {1, 2, 3, 4}  тоже что и a.union(b) 
print(a & b) # {2, 3}        тоже что и a.intersection(b)
print(a ^ b) # {1, 4}        тоже что и a.symmetric_difference(b)
print(a - b) # {1}           тоже что и a.difference(b)
print(a <= b) # False        тоже что и a.issubset(b)
print(a < (b | a)) # True    {1, 2, 3} вложено в {1, 2, 3, 4} 
print(a > a) # False         так как a = a
```

Также по аналогии с этими операциями, есть операции изменяющие сами множества (к функциям нужно добавить приставочку `_update`):

```python
a = {1, 2, 3}
b = {2, 3, 4}
b.difference_update(a) # b = {4}   тоже что и b -= a
a |= b  # a = {1, 2, 3, 4}         тоже что и a.update(b)
print(a, b)
```

## Dict

Словарь - множество пар $(\text{ключ} \mapsto \text{значение})$, ключи уникальны, значения могут повторяться. Иными словами перед нами отображение. Чтобы по ключу найти значение, в которое этот ключ переходит, можно обратиться к словарю "по индексу". 

```python
student = {
    "name": "Petr",
    "age": 16,
    "grade": 10,
    100: 67
}
print(student["name"], student[100]) 
```

Обращаться по индексу можно только к тем ключам, для которых словарь задан:

```python
student[30]
```

Для новых ключей значение можно задать так: 

```python
student[30] = 60
```

Создать словарь можно следующим образом:

```python
d1 = {}
d2 = dict()
d3 = dict([("a", 1), ("b", 2), ("c", 3)])
d4 = {x : x*x for x in range(1, 10) if x % 2 == 1}

print(d1, d2, d3, d4)
```

Обратите внимание, что `{}` и `{1:1}` - словари, а `{1, 1}`  множество.

| Методы со словарями | Что делают |
|------|-----|
|`len(d)`|кол-во элементов в `d`|
|`d[key]`|возвращает значение по ключу (если его нет - ошибка)|
|`d.get(key)`| возвращает значение по ключу `key` (если его нет - возвращает `None`)|
|`d.get(key, e)`| возвращает значение по ключу `key` (если его нет - возвращает `e`)|
|`del d[key]`|удаляет ключ `key` (ключ должен быть)|
|`d.pop(key)`|удаляет ключ `key` (ключ должен быть), возвращает значение по `key`|
|`key in d`| проверяет, что `key` есть в словаре|
|`d.keys()`| возвращает набор всех ключей словаря (типа `dict_keys`, можно привести к списку через `list`)|
|`d.values()`| возвращает набор всех значений c повторениями (типа `dict_values`, можно привести к списку через `list`)|
|`d.items()`| возвращает набор кортежей (ключ, значение) (типа `dict_items`, можно привести к списку через `list`)|
|`d1 \|= d2` / `d1.update(d2)`| добавить в словарь `d1` ключи и значения из `d2` (если в `d1` ключ уже определен, его значение все равно поменяется на значение из `d2`)|

```python
print(d3.keys()) # dict_keys(['a', 'b', 'c'])
print(d3.values()) # dict_values([1, 2, 3])
print(d3.items()) # dict_items([('a', 1), ('b', 2), ('c', 3)])
d3[1] = d3.get(1, 5) + 1
for key in d3:
    print(key, d3[key], sep = ": ")
print()
for key, value in d3.items():
    print(key, value, sep = "->")
```

```python
d3.update({1: 1, 2: 4})
print(d3.pop('b'))
print(d3)
```

## Хеширование

В качестве элементов `set` и ключей `dict` могут выступать не любые объекты, а только **неизменяемые хешируемые объекты**. Давайте разберем, как это работает:

Объекты хешируемы, значит для них определена **хеш-функция** `hash`, которая по нашему объекту возвращает некоторое число. При этом необходимо, чтобы $a = b \Rightarrow hash(a) = hash(b)$. Заметьте, что в обратную сторону следствие не обязательно. Однако, чем лучше хеш-функция, тем меньше должно быть пар повторяющихся элементов.

Изначально на `set` и `dict` выделяется сколько-то памяти и создается таблица некоторого размера $N$. Когда мы добавляем элемент в `set`, ровно как и ключ в `dict` сначала для этого элемента вычисляется хеш-функция - некоторое число $h$, после чего элемент кладется в $h \% N$ столбец таблицы множества/словаря. В этот момент может возникнуть **коллизия**, так как возможно в этом столбце уже лежит число с тамим же хешом по модулю $N$. Бороться с этим можно несколькими способами, например есть алгоритм Кукушки, в котором мы начинаем перекладывать элемент в следующую ячейку, пока не найдем свободную. В `set` и `dict` борятся самым простым путем, кладут в соответствующий столбец несколько значений. Тогда, чтобы проверить что элемент лежит в множестве или удалить элемент, мы должны посмотреть на столбец с соответсвующим хешом, пройтись по всем элементам, которые находятся в этом столбце и найти/удалить необходимый элемент.

Теперь становится понятно, что плохо если у большого числа элементов хеши совпадут - тогда в одном столбце будет скопление элементов и искать среди них элемент будет дорого. Однако для хорошей хеш-функции числа распределены равномерно и с небольшой вероятностью будут возникать коллизии, примерно с вероятностью $1/N$, таким образом если $N$ примерно совпадает с количеством элементов, то в каждом столбце будет мало элементов, а соответственно все функции нахождения и удаления будут работать быстро.

Еще раз сформулируем требования к объектам `set` и `dict`: Для данных объектов должна вычисляться хеш-функция, которая для равных элементов всегда вычисляет один и тот же хеш, при этом, чтобы операции работали быстро, для разных элементов хеши должны совпадать с маленькой вероятностью - значения хеш-функции должны быть равномерны.

Помимо этого объект должен быть неизменяемый, так как иначе в случае изменения объекта, его хеш может поменяться, а тогда найти новый объект в столбце с новым хешом не получится. В качетсве таких объектов выступают числа, строки, кортежи, но список, например, нельзя положить в `set` и `dict`.

```python
print(hash(5))           # 5
print(hash('hello'))     # 1461675741933155544
print(hash((1.4, 'hi'))) # -3048543930626743189
```

```python
hash([1, 2, 3])
```

```python
set([5, 'hello', (1.4, 'hi')])
```

```python
set([[1, 2, 3]])
```

Хеширование позволяет `set` и `dict` быстро вычислять операции с элементами за счет быстрого вычисления хеш-функции и редких коллизий.

```python
import time

n = 50000
a = set(range(n))
start = time.time()
cnt = 0
for i in range(n):
    if i in a:
        cnt += 1
print(n, "operations with set, time:", time.time() - start)

a = list(range(n))
start = time.time()
cnt = 0
for i in range(n):
    if i in a:
        cnt += 1
print(n, "operations with list, time:", time.time() - start)
```

Даже для не очень большого количества операций разница существенна.

Можно задать свой тип с хешом, который можно будет класть в `set` и `dict`, для этого нужно задать как минимум три функции `__init__`, которая сознает объект класса, `__hash__`, которая возвращает хеш объекта, и функция `__eq__`, которая проверяет, равны ли объекты.

```python
a = {1, 2, 3}
b = {1, 2}
c = {1}
d = {3, 2, 1}
print({a, b, c, d}) # не работает, потому что set не имеет хеш-функции
```

```python
class MySet:
    def __init__(self, s):
        self.set = s

    def __hash__(self):
        h = 0
        for x in self.set:
            h += hash(x)
        h %= 10**9
        return h
    
    def __eq__(self, other):
        return isinstance(other, MySet) and self.set == other.set

    def __repr__(self):
        return f"{self.set}"


a = MySet({1, 2, 3})
b = MySet({1, 2})
c = MySet({1})
d = MySet({3, 2, 1})
print({a, b, c, d})
```

Функция `isinstance(a, type)` проверяет, что объект `a` принимает тип `type`, функция `__repr__` нужна, чтобы можно было выводить данные. Приведем пример класса без `__repr__`:

```python
class A:
    def __init__(self, x):
        self.x = x

a = A(1)
b = A(2)
print(a, b.x)
```

Без `__repr__` функция `print` не отображает объект нормально `<__main__.A object...>`, стоит добавить функцию `__repr__` и вывод становится осмысленным:

```python
class A:
    def __init__(self, x):
        self.x = x
    def __repr__(self):
        return "A object with x = " + str(self.x)

a = A(1)
print(a)
```
